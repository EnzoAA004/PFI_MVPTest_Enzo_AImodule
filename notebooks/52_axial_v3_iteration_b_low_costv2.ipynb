# Axial v3 — Iteración B / B0 con I/O DICOM versionado

Este notebook vuelve a ejecutar **B0 completo sobre train/validation** con el soporte `.ima`/`.dcm` incluido en el commit del repositorio.

Controles incorporados:

- checkout exacto de un commit reproducible;
- prueba automática del loader DICOM versionado;
- comprobación sobre un `.ima` real del dataset;
- exclusión del split test;
- archivo automático del B0 provisional anterior;
- `RESUME_MODE=never` para evitar reutilizar el checkpoint provisional;
- preflight antes del entrenamiento completo;
- registro final de procedencia del runtime.


```python
# ============================================================
# 1) Dependencias, Drive, configuración y checkout reproducible
# ============================================================

import hashlib
import importlib.util
import json
import os
import shutil
import subprocess
import sys
import time
from pathlib import Path

REQUIRED_PACKAGES = {
    "numpy": "numpy",
    "pandas": "pandas",
    "PIL": "Pillow",
    "matplotlib": "matplotlib",
    "torch": "torch",
    "pydicom": "pydicom>=2.4",
    "pytest": "pytest>=8.0",
}

missing = [
    pip_package
    for module_name, pip_package in REQUIRED_PACKAGES.items()
    if importlib.util.find_spec(module_name) is None
]

if missing:
    subprocess.check_call([
        sys.executable,
        "-m",
        "pip",
        "install",
        "--quiet",
        *missing,
    ])

import numpy as np
import pandas as pd
import torch

IN_COLAB = "COLAB_RELEASE_TAG" in os.environ

if not IN_COLAB:
    raise RuntimeError(
        "Este notebook está preparado para Google Colab."
    )

from google.colab import drive

DRIVE_MOUNT = Path("/content/drive")

# Fuerza una nueva autenticación para elegir la cuenta secundaria.
drive.mount(
    str(DRIVE_MOUNT),
    force_remount=True,
)

MY_DRIVE = DRIVE_MOUNT / "MyDrive"
SHORTCUT_TARGETS = (
    DRIVE_MOUNT
    / ".shortcut-targets-by-id"
)

if not MY_DRIVE.is_dir():
    raise FileNotFoundError(
        f"No se pudo montar Mi unidad: {MY_DRIVE}"
    )


# ------------------------------------------------------------
# Localizar PFI_MVP
# ------------------------------------------------------------

pfi_candidates = []

# Caso normal: acceso directo agregado en la raíz de Mi unidad.
direct_candidate = MY_DRIVE / "PFI_MVP"

if direct_candidate.is_dir():
    pfi_candidates.append(
        direct_candidate
    )

# Buscar hasta tres niveles dentro de Mi unidad.
for current_root, directories, _ in os.walk(
    MY_DRIVE
):
    current_path = Path(current_root)

    try:
        depth = len(
            current_path
            .relative_to(MY_DRIVE)
            .parts
        )
    except ValueError:
        continue

    if depth >= 3:
        directories[:] = []
        continue

    for directory_name in directories:
        if directory_name.lower() == "pfi_mvp":
            candidate = (
                current_path
                / directory_name
            )

            if candidate.is_dir():
                pfi_candidates.append(
                    candidate
                )

# Fallback para accesos directos de Google Drive.
if SHORTCUT_TARGETS.is_dir():
    for candidate in (
        SHORTCUT_TARGETS
        .glob("*/PFI_MVP")
    ):
        if candidate.is_dir():
            pfi_candidates.append(
                candidate
            )

# Eliminar duplicados conservando el orden.
unique_candidates = []

for candidate in pfi_candidates:
    resolved_text = str(candidate)

    if resolved_text not in {
        str(item)
        for item in unique_candidates
    }:
        unique_candidates.append(
            candidate
        )

if not unique_candidates:
    visible_directories = sorted(
        path.name
        for path in MY_DRIVE.iterdir()
        if path.is_dir()
    )

    raise FileNotFoundError(
        "No se encontró PFI_MVP dentro de Mi unidad.\n\n"
        "La carpeta compartida todavía está solamente en "
        "'Compartidos conmigo'.\n"
        "En Google Drive: tres puntos sobre PFI_MVP → "
        "Organizar → Añadir acceso directo → Mi unidad.\n\n"
        f"Carpetas visibles en Mi unidad: {visible_directories}"
    )

PFI_ROOT = unique_candidates[0]

print({
    "pfiRoot": str(PFI_ROOT),
    "pfiRootExists": PFI_ROOT.is_dir(),
    "allCandidates": [
        str(candidate)
        for candidate in unique_candidates
    ],
})


# ------------------------------------------------------------
# Commit y repositorio
# ------------------------------------------------------------

REPO_REF = (
    "ae74b3b62ecad52dc20b22a76584fc107e609a1e"
)

REPO_BRANCH = (
    "fix/axial-v3-dicom-io-20260724"
)

REPO_ROOT = Path(
    "/content/PFI_MVPTest_Enzo_AImodule"
)

REPO_URL = (
    "https://github.com/EnzoAA004/"
    "PFI_MVPTest_Enzo_AImodule.git"
)


# ------------------------------------------------------------
# Rutas del proyecto
# ------------------------------------------------------------

MANIFEST_PATH = (
    PFI_ROOT
    / "results"
    / "E9_alkafri_axial_t2_final_labels_baseline"
    / "E9_t2_final_labels_curated_split.csv"
)

OUTPUT_BASE = (
    PFI_ROOT
    / "outputs"
)

ITERATION_B_ROOT = (
    OUTPUT_BASE
    / "axial_v3"
    / "iteration_b"
)

RUN_ID = "axial-v3-B0"

RUN_DIR = (
    ITERATION_B_ROOT
    / RUN_ID
)

REGISTRY_PATH = (
    ITERATION_B_ROOT
    / "experiment_registry.csv"
)

ARCHIVE_ROOT = (
    ITERATION_B_ROOT
    / "archive"
    / "provisional_unversioned_dicom"
)


# ------------------------------------------------------------
# Configuración de ejecución
# ------------------------------------------------------------

RUN_MODE_VALUE = "train"
EXPERIMENT_ID_VALUE = "B0"
AUTHORIZE_EXECUTION = True
ARCHIVE_EXISTING_B0 = True

os.environ["PFI_REPO_REF"] = (
    REPO_REF
)

os.environ["PFI_REPO_ROOT"] = str(
    REPO_ROOT
)

os.environ["PFI_REPO_URL"] = (
    REPO_URL
)

os.environ["PFI_DATASET_ROOT"] = str(
    PFI_ROOT
)

os.environ[
    "AXIAL_E9_CURATED_SPLIT_CSV"
] = str(
    MANIFEST_PATH
)

os.environ["PFI_OUTPUT_ROOT"] = str(
    OUTPUT_BASE
)

os.environ["PFI_RESUME_ROOT"] = str(
    OUTPUT_BASE
)

os.environ["PFI_REGISTRY_PATH"] = str(
    REGISTRY_PATH
)

os.environ["RUN_MODE"] = (
    RUN_MODE_VALUE
)

os.environ[
    "PFI_AXIAL_V3_EXPERIMENT_ID"
] = EXPERIMENT_ID_VALUE

os.environ[
    "PFI_AXIAL_V3_EXPERIMENT_TYPE"
] = "B0"

os.environ[
    "PFI_RUN_AXIAL_V3_EXPERIMENT"
] = (
    "1"
    if AUTHORIZE_EXECUTION
    else "0"
)

os.environ["PFI_MASK_LABEL_MODE"] = (
    "raw"
)

os.environ["PFI_NUM_WORKERS"] = (
    "0"
)

os.environ["PFI_BATCH_SIZE"] = (
    "8"
)

os.environ["PFI_USE_AMP"] = (
    "1"
)

os.environ["RESUME_MODE"] = (
    "never"
)

os.environ["PFI_AI_SERVICE_COMMIT"] = (
    "285159982832abb604a176b4302ac83a837ff1c9"
)


# ------------------------------------------------------------
# Validaciones
# ------------------------------------------------------------

if not MANIFEST_PATH.is_file():
    raise FileNotFoundError(
        "Se encontró PFI_MVP, pero no el manifest:\n"
        f"{MANIFEST_PATH}"
    )

if not torch.cuda.is_available():
    raise RuntimeError(
        "B0 completo requiere GPU CUDA. "
        "En Colab: Entorno de ejecución → "
        "Cambiar tipo de entorno → T4/L4."
    )


# ------------------------------------------------------------
# Checkout reproducible
# ------------------------------------------------------------

if not (REPO_ROOT / ".git").exists():
    subprocess.check_call([
        "git",
        "clone",
        REPO_URL,
        str(REPO_ROOT),
    ])

subprocess.check_call([
    "git",
    "-C",
    str(REPO_ROOT),
    "fetch",
    "--all",
    "--tags",
])

subprocess.check_call([
    "git",
    "-C",
    str(REPO_ROOT),
    "checkout",
    REPO_REF,
])

effective_sha = (
    subprocess.check_output(
        [
            "git",
            "-C",
            str(REPO_ROOT),
            "rev-parse",
            "HEAD",
        ],
        text=True,
    )
    .strip()
)

if effective_sha != REPO_REF:
    raise RuntimeError(
        "Commit efectivo inesperado: "
        f"{effective_sha} != {REPO_REF}"
    )

SRC_ROOT = (
    REPO_ROOT
    / "src"
)

if str(SRC_ROOT) not in sys.path:
    sys.path.insert(
        0,
        str(SRC_ROOT),
    )


def sha256_file(
    path: Path,
) -> str:
    digest = hashlib.sha256()

    with path.open("rb") as handle:
        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(chunk)

    return digest.hexdigest()


print({
    "repoBranch": REPO_BRANCH,
    "effectiveSha": effective_sha,
    "pfiRoot": str(PFI_ROOT),
    "manifestExists": (
        MANIFEST_PATH.is_file()
    ),
    "cudaAvailable": (
        torch.cuda.is_available()
    ),
    "deviceName": (
        torch.cuda.get_device_name(0)
    ),
    "runMode": RUN_MODE_VALUE,
    "experimentId": (
        EXPERIMENT_ID_VALUE
    ),
    "resumeMode": (
        os.environ["RESUME_MODE"]
    ),
})
```

```python
# ============================================================
# 2) Prueba automática y prueba sobre un .ima real
# ============================================================

# Prueba comprometida en el mismo commit que se ejecutará.
subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pytest",
        "-q",
        "tests/test_axial_v3_medical_io.py",
    ],
    cwd=str(REPO_ROOT),
)

import lumbar_mri.axial_v3.training_medical as training

training.assert_medical_io_installed()

if training.open_2d_array.__module__ != "lumbar_mri.axial_v3.medical_io":
    raise RuntimeError(
        "El loader activo no proviene del módulo versionado medical_io."
    )

manifest_df = pd.read_csv(MANIFEST_PATH)

split_counts = (
    manifest_df["split"]
    .astype(str)
    .str.lower()
    .value_counts()
    .to_dict()
)

expected_splits = {
    "train": 427,
    "test": 102,
    "val": 81,
}

if split_counts != expected_splits:
    raise RuntimeError(
        f"Distribución inesperada del manifest: {split_counts}"
    )

development_df = manifest_df[
    manifest_df["split"]
    .astype(str)
    .str.lower()
    .isin(["train", "val", "validation"])
]

sample_dicom_path = None
for raw_path in development_df["image_file_path"].astype(str):
    candidate = Path(raw_path)
    if not candidate.is_absolute():
        candidate = PFI_ROOT / candidate
    if candidate.suffix.lower() in {".ima", ".dcm"}:
        sample_dicom_path = candidate
        break

if sample_dicom_path is None:
    raise RuntimeError("No se encontró un DICOM en train/validation.")

sample_array = training.open_2d_array(sample_dicom_path)
medical_io_source = (
    REPO_ROOT
    / "src"
    / "lumbar_mri"
    / "axial_v3"
    / "medical_io.py"
)

print({
    "targetedTests": "passed",
    "loaderModule": training.open_2d_array.__module__,
    "medicalIoSource": str(medical_io_source),
    "medicalIoSourceSha256": sha256_file(medical_io_source),
    "manifestRows": len(manifest_df),
    "splitCounts": split_counts,
    "developmentRows": len(development_df),
    "sampleDicom": str(sample_dicom_path),
    "sampleShape": list(sample_array.shape),
    "sampleDtype": str(sample_array.dtype),
    "sampleFinite": bool(np.isfinite(sample_array).all()),
})

if sample_array.ndim != 2:
    raise RuntimeError(f"El DICOM no quedó 2D: {sample_array.shape}")
```

```python
# ============================================================
# 3) Archivar el B0 provisional anterior
# ============================================================

CURRENT_PROVENANCE_PATH = RUN_DIR / "runtime_provenance.json"

if CURRENT_PROVENANCE_PATH.is_file():
    existing_provenance = json.loads(
        CURRENT_PROVENANCE_PATH.read_text(encoding="utf-8")
    )
    if existing_provenance.get("effectiveSha") == REPO_REF:
        raise RuntimeError(
            "Ya existe un B0 completo generado con este mismo commit. "
            "No se vuelve a ejecutar automáticamente."
        )

archive_record = {
    "archived": False,
    "sourceRunDir": str(RUN_DIR),
    "archiveDir": None,
    "registrySnapshot": None,
}

if RUN_DIR.exists():
    if not ARCHIVE_EXISTING_B0:
        raise RuntimeError(
            f"Ya existe {RUN_DIR}. "
            "Activar ARCHIVE_EXISTING_B0 para preservar y reemplazar."
        )

    timestamp = time.strftime(
        "%Y%m%dT%H%M%SZ",
        time.gmtime(),
    )
    archive_dir = ARCHIVE_ROOT / f"{RUN_ID}_{timestamp}"
    archive_dir.parent.mkdir(parents=True, exist_ok=True)

    shutil.move(
        str(RUN_DIR),
        str(archive_dir),
    )

    registry_snapshot = None
    if REGISTRY_PATH.is_file():
        registry_snapshot = (
            archive_dir
            / f"experiment_registry_before_rerun_{timestamp}.csv"
        )
        shutil.copy2(
            REGISTRY_PATH,
            registry_snapshot,
        )

    archive_record = {
        "archived": True,
        "sourceRunDir": str(RUN_DIR),
        "archiveDir": str(archive_dir),
        "registrySnapshot": (
            str(registry_snapshot)
            if registry_snapshot is not None
            else None
        ),
    }

print(archive_record)
```

```python
# ============================================================
# Reparar la ruta real de PFI_MVP en una carpeta compartida
# ============================================================

import os
from pathlib import Path

EXPECTED_IMAGE = "T2_TSE_TRA__0001_003.ima"
EXPECTED_MASK = "L1_0001_D3.png"

SEARCH_ROOTS = [
    Path("/content/drive/MyDrive"),
    Path("/content/drive/.shortcut-targets-by-id"),
]

def find_exact_file(filename: str):
    matches = []

    for search_root in SEARCH_ROOTS:
        if not search_root.is_dir():
            continue

        for current_root, directories, files in os.walk(
            search_root
        ):
            if filename in files:
                matches.append(
                    Path(current_root) / filename
                )

                # Un resultado válido alcanza.
                return matches[0]

    return None


image_found = find_exact_file(
    EXPECTED_IMAGE
)

mask_found = find_exact_file(
    EXPECTED_MASK
)

print({
    "imageFound": (
        str(image_found)
        if image_found
        else None
    ),
    "maskFound": (
        str(mask_found)
        if mask_found
        else None
    ),
})

if image_found is None or mask_found is None:
    raise FileNotFoundError(
        "El acceso compartido permite ver el manifest, "
        "pero no permite acceder a todo PFI_MVP/data. "
        "Comprobá desde la cuenta propietaria que la carpeta "
        "PFI_MVP completa —incluida data— esté compartida, "
        "y no solamente results/outputs o una carpeta parcial."
    )


# La ruta buscada debe contener:
# PFI_MVP/data/AXIAL_ALKAFRI/...
def infer_pfi_root(path: Path) -> Path:
    parts = path.parts

    for index, part in enumerate(parts):
        if part == "PFI_MVP":
            return Path(*parts[: index + 1])

    # Fallback: buscar el ancestro inmediatamente anterior a data.
    for parent in path.parents:
        if parent.name == "data":
            return parent.parent

    raise RuntimeError(
        f"No se pudo inferir PFI_ROOT desde {path}"
    )


IMAGE_PFI_ROOT = infer_pfi_root(
    image_found
)

MASK_PFI_ROOT = infer_pfi_root(
    mask_found
)

if IMAGE_PFI_ROOT != MASK_PFI_ROOT:
    raise RuntimeError(
        "Imagen y máscara aparecen bajo raíces diferentes:\n"
        f"imagen: {IMAGE_PFI_ROOT}\n"
        f"máscara: {MASK_PFI_ROOT}"
    )

PFI_ROOT = IMAGE_PFI_ROOT

MANIFEST_PATH = (
    PFI_ROOT
    / "results"
    / "E9_alkafri_axial_t2_final_labels_baseline"
    / "E9_t2_final_labels_curated_split.csv"
)

OUTPUT_BASE = (
    PFI_ROOT
    / "outputs"
)

ITERATION_B_ROOT = (
    OUTPUT_BASE
    / "axial_v3"
    / "iteration_b"
)

RUN_ID = "axial-v3-B0"

RUN_DIR = (
    ITERATION_B_ROOT
    / RUN_ID
)

REGISTRY_PATH = (
    ITERATION_B_ROOT
    / "experiment_registry.csv"
)

ARCHIVE_ROOT = (
    ITERATION_B_ROOT
    / "archive"
    / "provisional_unversioned_dicom"
)

os.environ["PFI_DATASET_ROOT"] = str(
    PFI_ROOT
)

os.environ[
    "AXIAL_E9_CURATED_SPLIT_CSV"
] = str(
    MANIFEST_PATH
)

os.environ["PFI_OUTPUT_ROOT"] = str(
    OUTPUT_BASE
)

os.environ["PFI_RESUME_ROOT"] = str(
    OUTPUT_BASE
)

os.environ["PFI_REGISTRY_PATH"] = str(
    REGISTRY_PATH
)

print({
    "repairedPfiRoot": str(PFI_ROOT),
    "manifestExists": (
        MANIFEST_PATH.is_file()
    ),
    "imageExists": (
        image_found.is_file()
    ),
    "maskExists": (
        mask_found.is_file()
    ),
    "outputBase": str(OUTPUT_BASE),
})
```

```python
# ============================================================
# 4) Construir B0, ejecutar preflight y entrenar desde cero
# ============================================================

from dataclasses import replace

from lumbar_mri.axial_v3.experiments import (
    estimate_run_count,
    load_low_cost_grid,
    select_experiment,
)
from lumbar_mri.axial_v3.guards import require_train_val_only

ALLOWED_SPLITS = require_train_val_only(
    ["train", "val"],
    context="notebook_52_versioned_dicom",
)

CONFIG_PATH = (
    REPO_ROOT
    / "config"
    / "axial_v3_low_cost_grid.json"
)
GRID = load_low_cost_grid(CONFIG_PATH)

def train_config_for_experiment(experiment_id: str):
    base = training.AxialV3TrainConfig.from_env()
    experiment = select_experiment(GRID, experiment_id)

    return replace(
        base,
        RUN_ID=experiment.runId,
        EXPERIMENT_ID=experiment.experimentId,
        EXPERIMENT_TYPE=experiment.experimentType,
        RAW0_WEIGHT_BOOST=experiment.raw0EffectiveWeightMultiplier,
        MAX_CLASS_WEIGHT_RATIO=experiment.maxClassWeightRatio,
        LOSS_NAME=experiment.lossName,
        TVERSKY_ALPHA=(
            experiment.tverskyAlpha
            if experiment.tverskyAlpha is not None
            else base.TVERSKY_ALPHA
        ),
        TVERSKY_BETA=(
            experiment.tverskyBeta
            if experiment.tverskyBeta is not None
            else base.TVERSKY_BETA
        ),
        FOCAL_GAMMA=(
            experiment.focalGamma
            if experiment.focalGamma is not None
            else base.FOCAL_GAMMA
        ),
        RAW0_TVERSKY_WEIGHT=experiment.raw0TverskyWeight,
        RAW0_FP_PENALTY_WEIGHT=experiment.raw0FpPenaltyWeight,
        MIN_PROBABILITY=experiment.minProbability,
        MIN_MARGIN=experiment.minMargin,
        MAX_EPOCHS=experiment.maxEpochs,
        EARLY_STOPPING_PATIENCE=experiment.earlyStoppingPatience,
        PRESENCE_HEAD_ENABLED=experiment.presenceHeadEnabled,
        LAMBDA_PRESENCE=experiment.lambdaPresence,
        RAW0_BALANCED_SAMPLER_ENABLED=(
            experiment.raw0BalancedSamplerEnabled
        ),
        POSITIVE_FRACTION=(
            experiment.positiveFraction
            if experiment.positiveFraction is not None
            else 0.5
        ),
        PARENT_EXPERIMENT_ID=experiment.parentExperimentId,
        PARENT_RUN_ID=experiment.parentRunId,
        RESUME_MODE="never",
    )

CFG = train_config_for_experiment("B0")
EXPERIMENT = select_experiment(GRID, "B0")

print(json.dumps(
    {
        "runId": CFG.RUN_ID,
        "experimentId": CFG.EXPERIMENT_ID,
        "experimentType": CFG.EXPERIMENT_TYPE,
        "resumeMode": CFG.RESUME_MODE,
        "manifest": str(CFG.SPLIT_MANIFEST_PATH),
        "outputRoot": str(CFG.OUTPUT_ROOT),
        "registryPath": str(CFG.REGISTRY_PATH),
        "plannedRuns": estimate_run_count(GRID),
        "gitCommit": effective_sha,
        "aiServiceCommit": os.environ["PFI_AI_SERVICE_COMMIT"],
    },
    indent=2,
))

preflight = training.run_preflight(
    CFG,
    EXPERIMENT,
)

print(json.dumps(
    {
        "preflightStatus": preflight.get("status"),
        "trainSlices": preflight.get("trainSlices"),
        "valSlices": preflight.get("valSlices"),
        "trainPatients": preflight.get("trainPatients"),
        "valPatients": preflight.get("valPatients"),
        "device": preflight.get("device"),
        "ampEnabled": preflight.get("ampEnabled"),
        "runtimeShape": preflight.get("runtimeShape"),
    },
    indent=2,
    default=str,
))

if preflight.get("status") != "preflight_passed":
    raise RuntimeError("El preflight versionado de B0 no fue aprobado.")

result = training.run_training(
    CFG,
    smoke=False,
)
```

```python
# ============================================================
# 5) Validación final y procedencia
# ============================================================

if result.get("smokeOnly") is not False:
    raise RuntimeError("La corrida terminada no fue un entrenamiento completo.")

runtime_smoke = result.get("runtimeSmoke") or {}
if runtime_smoke.get("finite") is not True:
    raise RuntimeError("La salida de runtime no es finita.")

resume_status = result.get("resumeStatus") or {}
if resume_status.get("loaded") is not False:
    raise RuntimeError("B0 reutilizó un checkpoint cuando debía iniciar desde cero.")

final_metrics = result.get("finalValidationMetrics") or {}

provenance = {
    "schemaVersion": "axial-v3-b0-versioned-dicom-v1",
    "effectiveSha": effective_sha,
    "repoBranch": REPO_BRANCH,
    "aiServiceCommit": os.environ["PFI_AI_SERVICE_COMMIT"],
    "medicalIoModule": training.open_2d_array.__module__,
    "medicalIoSource": str(medical_io_source),
    "medicalIoSourceSha256": sha256_file(medical_io_source),
    "manifestPath": str(MANIFEST_PATH),
    "manifestSha256": sha256_file(MANIFEST_PATH),
    "runId": result.get("runId"),
    "experimentId": result.get("experimentId"),
    "checkpointPath": result.get("checkpointPath"),
    "checkpointSha256": result.get("checkpointSha256"),
    "archiveRecord": archive_record,
}

RUN_DIR.mkdir(parents=True, exist_ok=True)
CURRENT_PROVENANCE_PATH.write_text(
    json.dumps(
        provenance,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)

summary = {
    "status": "B0_versioned_dicom_completed",
    "effectiveSha": effective_sha,
    "runId": result.get("runId"),
    "experimentId": result.get("experimentId"),
    "smokeOnly": result.get("smokeOnly"),
    "selectedEpoch": result.get("selectedEpoch"),
    "bestValidationMetric": result.get("bestValidationMetric"),
    "diceMacroForeground": final_metrics.get(
        "dice_macro_foreground"
    ),
    "diceMacroExcludingRaw0": final_metrics.get(
        "dice_macro_excluding_raw0"
    ),
    "raw0Dice": final_metrics.get("raw0Dice"),
    "raw0Precision": final_metrics.get("raw0Precision"),
    "raw0Recall": final_metrics.get("raw0Recall"),
    "raw0FalsePositivePixels": final_metrics.get(
        "raw0FalsePositivePixels"
    ),
    "raw0PredictedInGtAbsentCases": final_metrics.get(
        "raw0PredictedInGtAbsentCases"
    ),
    "raw0GtAbsentCases": final_metrics.get(
        "raw0GtAbsentCases"
    ),
    "durationSeconds": result.get("durationSeconds"),
    "checkpointPath": result.get("checkpointPath"),
    "checkpointSha256": result.get("checkpointSha256"),
    "lastCheckpointPath": result.get("lastCheckpointPath"),
    "lastCheckpointSha256": result.get(
        "lastCheckpointSha256"
    ),
    "resumeStatus": resume_status,
    "runtimeSmoke": runtime_smoke,
    "provenancePath": str(CURRENT_PROVENANCE_PATH),
}

summary_path = RUN_DIR / "reports" / "versioned_dicom_summary.json"
summary_path.parent.mkdir(parents=True, exist_ok=True)
summary_path.write_text(
    json.dumps(
        summary,
        indent=2,
        ensure_ascii=False,
    ),
    encoding="utf-8",
)

print(json.dumps(
    summary,
    indent=2,
    default=str,
))
```
